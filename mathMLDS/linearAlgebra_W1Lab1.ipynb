```python
a1 = np.array([[1, 1], [2, 2]])
a2 = np.array([[3, 3], [4, 4]])

print(f"a1:\n{a1}")
print(f"a2:\n{a2}")
```

```python
horz_stack = np.hstack((a1, a2))
print(horz_stack)
```
